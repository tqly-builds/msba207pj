```python
%matplotlib inline
from pathlib import Path
import numpy as np
import pandas as pd
from sklearn.linear_model import LogisticRegression
from sklearn.model_selection import train_test_split
import statsmodels.api as sm

DATA = Path('/user/data')

bank_df = pd.read_csv('UniversalBank.csv')
bank_df.drop(columns=['ID', 'ZIP Code'], inplace=True)
bank_df.columns = [c.replace(' ', '_') for c in bank_df.columns]

bank_df['Education'] = bank_df['Education'].astype('category')
new_categories = {1: 'Undergrad', 2: 'Graduate', 3: 'Advanced/Professional'}
bank_df.Education.cat.rename_categories(new_categories)

bank_df = pd.get_dummies(bank_df, prefix_sep='_', drop_first=True, dtype=int)

y = bank_df['Personal_Loan']
X = bank_df.drop(columns=['Personal_Loan'])

train_X, valid_X, train_y, valid_y = train_test_split(
    X, y, test_size=0.4, random_state=1
)
```

```python
from sklearn.linear_model import LogisticRegression
from sklearn.metrics import accuracy_score
```

```python
# Version 1
v1 = LogisticRegression(
    solver="liblinear",
    penalty="l2",
    C=1e42
)
v1.fit(train_X, train_y)

print("V 1 intercept:",v1.intercept_)
print()

coeff_v1 = pd.DataFrame({
    "Feature": train_X.columns,
    "Coefficient": v1.coef_[0]
})
print("v 1 coefficients: \n",coeff_v1)
```

```python
# Version 2
v2 = LogisticRegression(
    solver="liblinear",
    penalty="l2",
    C=1e42,
    tol=1e-10
)
v2.fit(train_X, train_y)

print("V 2 intercept:", v2.intercept_)
print()

coeff_v2 = pd.DataFrame({
    "Feature": train_X.columns,
    "Coefficient": v2.coef_[0]
})
print("v 2 coefficients: \n", coeff_v2)
```

```python
# Version 3
v3 = LogisticRegression(
    solver="lbfgs",
    penalty=None,
    max_iter=10000
)
v3.fit(train_X, train_y)

print("V 3 intercept:", v3.intercept_)
print()

coeff_v3 = pd.DataFrame({
    "Feature": train_X.columns,
    "Coefficient": v3.coef_[0]
})
print("v 3 coefficients: \n", coeff_v3)
```

```python
# Version 4
v4 = LogisticRegression(
    solver="lbfgs",
    penalty=None,
    max_iter=10000,
    tol=1e-28
)
v4.fit(train_X, train_y)

print("V 4 intercept:", v4.intercept_)
print()

coeff_v4 = pd.DataFrame({
    "Feature": train_X.columns,
    "Coefficient": v4.coef_[0]
})
print("v 4 coefficients: \n", coeff_v4)
```

```python
import statsmodels.api as sm

train_X_a = sm.add_constant(train_X)
glm = sm.GLM(train_y,train_X_a,family=sm.families.Binomial())
glm2=glm.fit()
print(f"AIC: {glm2.aic:.2f}")
print(glm2.summary())
```
